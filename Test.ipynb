## _Linear Search in List_

```python
index = -1
def linear_search(numbers, n):
    for i in range (len(numbers)):
        if numbers[i] == n:
            globals()['index'] = i
            return True
    return False

numbers = [4,8,5,9,1,2]
try:
    n = int(input("enter the element to search: "))
    
    if linear_search(numbers, n):
        print("element found at index ", index)
    else:        
        print("element not found") 
except ValueError:
    print("invalid input")
finally:
    print("")  
```

## _Binary Search in List_

```python
index = -1

def binary_search(numbers, n):
    low = 0; high = len(numbers) - 1

    while low <= high:
        mid = (low + high)//2
        if numbers[mid] == n:
            globals()['index'] = mid
            return True
        elif numbers[mid] > n:
            high = mid - 1
        elif numbers[mid] < n:
            low = mid + 1
        else:
            print("invalid input")

numbers = [2,5,8,13,24,56]

try:
    n = int(input("enter the element to search: "))

    if binary_search(numbers, n):
        print("element found at index ", index)
    else:        
        print("element not found") 
except ValueError:
    print("invalid input")
finally:
    print("") 
    
```
