## _Linear Search in List_

```python
index = -1
def search(numbers, n):
    for i in range (len(numbers)):
        if numbers[i] == n:
            globals()['index'] = i
            return True
    return False

numbers = [4,8,5,9,1,2]
n = int(input("enter an element to search: "))

if search(numbers, n):
    print("element found at index ", index)
else:
    print("element not found")    
```
