## _Linear Search in List_

```python
index = -1
def linear_search(numbers, n):
    for i in range (len(numbers)):
        if numbers[i] == n:
            globals()['index'] = i
            return True
    return False

numbers = [4,8,5,9,1,2]
try:
    n = int(input("enter the element to search: "))
    
    if linear_search(numbers, n):
        print("element found at index ", index)
    else:        
        print("element not found") 
except ValueError:
    print("invalid input")
finally:
    print("")  
```

## _Binary Search in List_

```python
index = -1

def binary_search(numbers, n):
    low = 0; high = len(numbers) - 1

    while low <= high:
        mid = (low + high)//2
        if numbers[mid] == n:
            globals()['index'] = mid
            return True
        elif numbers[mid] > n:
            high = mid - 1
        elif numbers[mid] < n:
            low = mid + 1
        else:
            print("invalid input")

numbers = [2,5,8,13,24,56]

try:
    n = int(input("enter the element to search: "))

    if binary_search(numbers, n):
        print("element found at index ", index)
    else:        
        print("element not found") 
except ValueError:
    print("invalid input")
finally:
    print("") 
    
```

## _Bubble Sort_

```python
def sorts(nums):
    for i in range (len(nums)-1,0,-1):
        for j in range (i):
            if nums[j] > nums[j+1]:
                temp = nums[j]
                nums[j] = nums[j+1]
                nums[j+1] = temp

nums = [5, 3, 8, 6, 7, 2]
print("unsorted: ",nums)
sorts(nums)

print("sorted: ",nums)
```

## _Selection Sort_

```python
def sort(nums):
    for i in range(5):
        minpos = i
        for j in range(i, 6):
            if nums[j] < nums[minpos]:
                minpos = j
        
        temp = nums[i]
        nums[i] = nums[minpos]
        nums[minpos] = temp
        print(nums)

nums = [4, 7, 2, 9, 5, 8]

print("unsorted:\n", nums)
sort(nums)
print("sorted:\n", nums)
```

```python
ls = [2,8,23,-65,6,9,-1,4,34]
for i in range (len(ls)):
    min = i
    for j in range (i, len(ls)):
        if ls[j] < ls[min]:
            min = j
    temp = ls[i]
    ls[i] = ls[min]
    ls[min] = temp
print(ls)
```

```python
# Reverse the given Array
# two pointer approach

arr = [1, 2, 3, 4, 5]

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1

print(arr)
```
