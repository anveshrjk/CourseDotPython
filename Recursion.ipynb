# **Recursion**

Q) print 1-5

```python
def print_numbers(L,R):
    # base case
    if L > R:
        return
    print_numbers(L+1, R)
    print(L, end=" ")
    
print_numbers(1, 5)
```

Q) Sum of array

```python
def iterative(arr):
    if not arr:
        return 0
    return arr[0] + iterative(arr[1:])

iterative([1,2,3,4,5])
```

Q) Factorial

```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n-1)
factorial(0)
```
