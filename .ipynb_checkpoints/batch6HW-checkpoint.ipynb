```python
# 1. Write a program using a for loop to print all Armstrong numbers between 100 and 999. (Armstrong number: sum of cubes of digits equals the number itself. Example: 153 => 1³+5³+3³ = 153).
try:
    for num in range(100, 1000):
        d1 = num // 100
        d2 = (num // 10) % 10
        d3 = num % 10
        if d1**3 + d2**3 + d3**3 == num:
            print(num)
except Exception as e:
    print("error")
```

```python
# 2. Write a program to generate and display the first n prime numbers using a for loop.
n = input("enter how many prime numbers to display: ")
try:
    n = int(n)
    count = 0
    num = 2
    while count < n:
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num)
            count += 1
        num += 1
except ValueError:
    print("invalid input")
```

```python
# 3. Write a program to display all numbers from 1 to 500 that are divisible by 3, but the sum of their digits should not exceed 10.
try:
    for num in range(1, 501):
        if num % 3 == 0:
            s = sum(int(d) for d in str(num))
            if s <= 10:
                print(num)
except Exception as e:
    print("error")
```

```python
''' 
4. Write a program using a for loop to print a pyramid of stars (*) of height n. Example for n=4:
   *
  ***
 *****
*******
'''
n = input("enter height: ")
try:
    n = int(n)
    for i in range(1, n + 1):
        print(' ' * (n - i) + '*' * (2 * i - 1))
except ValueError:
    print("invalid input")
```

```python
# 5. Write a program to accept a string and check whether it is a pangram (contains all 26 alphabets at least once) using a for loop.
text = input("enter text: ")
try:
    text = text.lower()
    found = ''
    for ch in text:
        if ch.isalpha() and ch not in found:
            found += ch
    if len(found) == 26:
        print("pangram")
    else:
        print("not pangram")
except Exception as e:
    print("error")
```

```python
# 6. Write a program using a for loop to print all twin primes between 1 and 100. (Twin primes: pairs of prime numbers with a difference of 2, e.g., (3,5), (11,13)).
try:
    for num in range(2, 99):
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            next_num = num + 2
            is_next_prime = True
            for j in range(2, int(next_num**0.5) + 1):
                if next_num % j == 0:
                    is_next_prime = False
                    break
            if is_next_prime and next_num <= 100:
                print(num, next_num)
except Exception as e:
    print("error")
```

```python
# 7. Write a program that accepts a number from the user and prints whether it is a Harshad number (number divisible by the sum of its digits) using a for loop.
num = input("enter a number: ")
try:
    num = int(num)
    s = 0
    for d in str(num):
        s += int(d)
    if num % s == 0:
        print("harshad number")
    else:
        print("not harshad number")
except ValueError:
    print("invalid input")
```

```python
# 8. Write a program to generate Pascal’s Triangle up to n rows using a for loop.
n = input("enter number of rows: ")
try:
    n = int(n)
    for i in range(n):
        print(' ' * (n - i), end='')
        val = 1
        for j in range(i + 1):
            print(val, end=' ')
            val = val * (i - j) // (j + 1)
        print()
except ValueError:
    print("invalid input")
```

10. Write a program that accepts a number from the user and prints whether it is a Strong number (sum of factorials of digits = number itself) using a for loop. Example: 145 => 1! + 4! + 5! = 145.


```python
num = input("enter a number: ")
try:
    num = int(num)
    s = 0
    for d in str(num):
        f = 1
        for i in range(1, int(d) + 1):
            f *= i
        s += f
    if s == num:
        print("strong number")
    else:
        print("not strong number")
except ValueError:
    print("invalid input")
```

##### *While Loop Problems:*

 11. Write a program using a while loop to find the reverse of a number and check if the reversed number is prime. Example: Input = 73 → Reverse = 37 → Prime.

```python
num = input("enter a number: ")
try:
    num = int(num)
    rev = 0
    temp = num
    while temp > 0:
        rev = rev * 10 + temp % 10
        temp //= 10
    print("reversed:", rev)
    if rev < 2:
        print("not prime")
    else:
        i = 2
        is_prime = True
        while i * i <= rev:
            if rev % i == 0:
                is_prime = False
                break
            i += 1
        if is_prime:
            print("prime")
        else:
            print("not prime")
except ValueError:
    print("invalid input")
```

 12. Write a program that continues to accept numbers from the user until the sum of digits of all numbers entered becomes greater than 100.

```python
total = 0
try:
    while total <= 100:
        num = input("enter a number: ")
        num = int(num)
        s = 0
        for d in str(abs(num)):
            s += int(d)
        total += s
    print("sum of digits exceeded 100")
except ValueError:
    print("invalid input")
```

 13. Write a program using a while loop to check whether a number is a Duck number (a number containing zero but not starting with zero, e.g., 202, 1203).

```python
num = input("enter a number: ")
try:
    if num[0] == '0':
        print("not duck number")
    else:
        i = 0
        found_zero = False
        while i < len(num):
            if num[i] == '0':
                found_zero = True
                break
            i += 1
        if found_zero:
            print("duck number")
        else:
            print("not duck number")
except Exception as e:
    print("invalid input")
```

 14. Write a program using a while loop to accept a number and check if it is a Happy number. (A number is happy if repeatedly replacing it with the sum of squares of its digits eventually reaches 1). Example: 19 is a happy number.


```python
num = input("enter a number: ")
try:
    num = int(num)
    seen = []
    while num != 1 and num not in seen:
        seen.append(num)
        s = 0
        temp = num
        while temp > 0:
            d = temp % 10
            s += d * d
            temp //= 10
        num = s
    if num == 1:
        print("happy number")
    else:
        print("not happy number")
except ValueError:
    print("invalid input")
```

 15. Write a program using a while loop to find the largest prime factor of a given number.


```python
num = input("enter a number: ")
try:
    num = int(num)
    i = 2
    largest = 0
    while i <= num:
        if num % i == 0:
            is_prime = True
            j = 2
            while j * j <= i:
                if i % j == 0:
                    is_prime = False
                    break
                j += 1
            if is_prime:
                largest = i
            num //= i
        else:
            i += 1
    print("largest prime factor:", largest)
except ValueError:
    print("invalid input")
```

17. Write a program using a while loop to compute the sum of digits of a number until the result becomes a single-digit number (Digital root). Example: 9875 => 9+8+7+5=29 => 2+9=11 => 1+1=2.


```python
# 17. Write a program using a while loop to compute the sum of digits of a number until the result becomes a single-digit number (Digital root). Example: 9875 => 9+8+7+5=29 => 2+9=11 => 1+1=2.
num = input("enter a number: ")
try:
    num = int(num)
    while num >= 10:
        s = 0
        while num > 0:
            s += num % 10
            num //= 10
        num = s
    print("digital root:", num)
except ValueError:
    print("invalid input")
```

```python
# 18. Write a program using a while loop to generate the Collatz sequence for a given number. (Rule: If n is even => n/2, if odd => 3n+1. Continue until n=1).
num = input("enter a number: ")
try:
    num = int(num)
    while num != 1:
        print(num)
        if num % 2 == 0:
            num //= 2
        else:
            num = 3 * num + 1
    print(num)
except ValueError:
    print("invalid input")
```

19. Write a program using a while loop to accept a number and check whether it is a Kaprekar number. (Kaprekar number: if square of the number can be split into two parts whose sum equals the number. Example: 45²=2025 => 20+25=45).


```python
num = input("enter a number: ")
try:
    num = int(num)
    sq = num * num
    str_sq = str(sq)
    found = False
    i = 1
    while i < len(str_sq):
        left = int(str_sq[:i])
        right = int(str_sq[i:])
        if right != 0 and left + right == num:
            found = True
            break
        i += 1
    if found or num == 1:
        print("kaprekar number")
    else:
        print("not kaprekar number")
except ValueError:
    print("invalid input")
```


20. Write a program to simulate an ATM machine using a while loop where a user can:
• Check balance  
• Deposit money  
• Withdraw money (only if balance is sufficient)  
• Exit  
Continue until the user chooses to exit.


```python

balance = 0
try:
    while True:
        print("\nchoose an option:")
        print("1. check balance")
        print("2. deposit money")
        print("3. withdraw money")
        print("4. exit")
        choice = input("enter your choice: ")

        if choice == '1':
            print("balance:", balance)
        elif choice == '2':
            amount = input("enter amount to deposit: ")
            amount = float(amount)
            if amount > 0:
                balance += amount
                print("amount deposited")
            else:
                print("invalid amount")
        elif choice == '3':
            amount = input("enter amount to withdraw: ")
            amount = float(amount)
            if amount > balance:
                print("insufficient balance")
            elif amount <= 0:
                print("invalid amount")
            else:
                balance -= amount
                print("amount withdrawn")
        elif choice == '4':
            print("thank you for using atm")
            break
        else:
            print("invalid choice")
except ValueError:
    print("invalid input")
```
