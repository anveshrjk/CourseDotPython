# NumPy

Its a multi-dimentional array library to store all sorts of data in 1D, 2D and 3D arrays.

#### Why NumPy over List
-> NumPy is faster than lists because its fixed in size and lists are dynamic in nature.  
-> It also allows Contigious memory allocation.

#### Benifits
-> SIMD Vector Processing unit can be utilized in NumPy because of contigious memory allocation (basically performs operations faster when memory is contigious).  
-> Effecient cache utilization.

#### Applications of NumPy
-> Mathematics (MATLAB replacement)  
-> Plotting  
-> Backend (Pandas, Connect 4, Digital Photography)
