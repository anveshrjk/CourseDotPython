Creating a database, a cursor and connecting it:

```python
import sqlite3

# con = sqlite3.connect(":memory:") this will store data directly in memory then gets deleted~

con = sqlite3.connect('customer.db')
cur = con.cursor()
```

Creating a table:

```python
cur.execute('''CREATE TABLE IF NOT EXISTS customers (
            first_name DATATYPE,
            last_name DATATYPE,
            email DATATYPE
            )
        ''')
```

Inserting values in the table

```python
cur.execute("INSERT OR IGNORE INTO customers VALUE ('Rose', 'Gurl', 'rosenotfound@gmail.com')")

'''
To insert multiple rows:
many_customers = [
                (),
                (),
                (),
                ... data in the list~
            ]
cur.executemany("INSERT OR IGNORE INTO customers VALUES                 (?,?,?)", many_customers)
'''
```

Query the database:

```python
# to show the table data:
cur.execute("SELECT * FROM customers") # ---> query to show table

items = cur.fetchall() # --- > fetch command to actually show
#cur.fetchone()[column]
#cur.fetchmany(x)

# print(items) <--- this will just print items in tuple format, which is not easily readable~

print(f"{'First Name':<12} {'Last Name':<12} {'Gmail':<30}")
print("-" * 54)
for item in items:
    print(f"{item[0]:<12} {item[1]:<12} {item[2]:<30}")
```

Commit and Close:

```python
# commit the command
con.commit()

# close the connection (con)
con.close()
```
